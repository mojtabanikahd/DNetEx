```python
def make_md_table_from_lists(numbers, means, stderrs, precision=3):
    header = "| Number | " + " | ".join(str(n) for n in numbers) + " |"
    separator = "| --- | " + " | ".join(["---"] * len(numbers)) + " |"
    cells = [f"{m:.{precision}f} ± {s:.{precision}f}" for m, s in zip(means, stderrs)]
    data_row = "| Value | " + " | ".join(cells) + " |"
    return "\n".join([header, separator, data_row])

# Example
numbers = ["01", "02", "04", "08", "16", "32"]
means = [runs_time_mean_list_01[0][0], runs_time_mean_list_02[0][0], runs_time_mean_list_04[0][0], runs_time_mean_list_08[0][0],
 runs_time_mean_list_16[0][0], runs_time_mean_list_32[0][0]]
stderrs = [runs_time_se_list_01[0][0], runs_time_se_list_02[0][0], runs_time_se_list_04[0][0], runs_time_se_list_08[0][0],
 runs_time_se_list_16[0][0], runs_time_se_list_32[0][0]]
print(make_md_table_from_lists(numbers, means, stderrs))
```
